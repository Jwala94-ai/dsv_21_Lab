### Q28. Trace conversion to dictionary and back to Series.

```python
s = pd.Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = pd.Series(d)
print("Dictionary:", d)
print("s2:\\n", s2)
print("Same labelled data:", s.equals(s2))
```

### Q29. Create a labelled product sales Series and perform all requested operations.

```python
products = pd.Series(
    [650, 420, 890, 350],
    index=['Laptop Stand', 'Mouse', 'Keyboard', 'Notebook'],
    name='Sales'
)
print("Values:", products.to_numpy())
print("Index:", products.index)
print("Keyboard sales:", products['Keyboard'])
print("Sales above 500:\\n", products[products > 500])
print("Does Mouse exist?", 'Mouse' in products.index)
product_dict = products.to_dict()
print("Dictionary:", product_dict)
```

### Q30. Analyse index order, NaN, filter, notnull, and names.

```python
data = {'A': 15, 'B': 35, 'C': 25}
s = pd.Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'

print("Series:\\n", s)
print("(a) Index order:", list(s.index))
print("(b) NaN label: D")
print("(c) Values greater than 20:\\n", s[s > 20])
print("(d) Non-missing flags:\\n", pd.notnull(s))
print("(e) Series name:", s.name)
print("Index name:", s.index.name)
```
