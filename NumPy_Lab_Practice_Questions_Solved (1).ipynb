### Q12. Demonstrate the difference between a view and a copy.

```python
original = np.array([10, 20, 30, 40, 50])
view = original[1:4]
copy = original[1:4].copy()

view[0] = 999
print("After changing view:")
print("Original:", original)
print("View:", view)

copy[1] = 777
print("\\nAfter changing copy:")
print("Original:", original)
print("Copy:", copy)
```

### Q20. Use transpose() and swapaxes() on a 3-D array.

```python
arr = np.arange(1, 25).reshape(2, 3, 4)
print("Original shape:", arr.shape)
transposed = arr.transpose(1, 0, 2)
print("transpose(1, 0, 2) shape:", transposed.shape)
swapped = np.swapaxes(arr, 0, 2)
print("swapaxes(0, 2) shape:", swapped.shape)
print("Transposed array:\\n", transposed)
print("Swapped array:\\n", swapped)
```

## 11. Integrated Lab Questions

### Q39. Student marks analysis: totals, averages, and students averaging ≥ 50.

```python

marks = np.array([
    [65, 70, 80],
    [35, 45, 50],
    [90, 85, 95],
    [40, 55, 45],
    [72, 68, 75],
    [20, 30, 25],
    [55, 60, 50],
    [88, 92, 84],
    [48, 52, 46],
    [76, 73, 79]
])
totals = np.sum(marks, axis=1)
averages = np.mean(marks, axis=1)
qualified = np.where(averages >= 50)[0]
print("Marks:\\n", marks)
print("Totals:", totals)
print("Averages:", averages)
print("Qualified student indices (0-based):", qualified)
print("Qualified student numbers (1-based):", qualified + 1)
```

### Q40. Matrix analysis for 1–16 in a 4 × 4 matrix.

```python
matrix = np.arange(1, 17).reshape(4, 4)
print("Matrix:\\n", matrix)
print("Transpose:\\n", matrix.T)
print("Row-wise sums:", np.sum(matrix, axis=1))
print("Column-wise sums:", np.sum(matrix, axis=0))
print("Mean:", np.mean(matrix))
print("Maximum:", np.max(matrix))
print("Minimum:", np.min(matrix))
print("Central 2 × 2:\\n", matrix[1:3, 1:3])
```

### Q41. Random data analysis with 20 observations from np.random.randn().

### Q42. Matrix operations challenge.

```python
A = np.arange(1, 10).reshape(3, 3)
B = np.array([[9, 8, 7], [6, 5, 4], [3, 2, 1]])
addition = A + B
elementwise = A * B
product = A @ B
transpose = A.T
print("A:\\n", A)
print("B:\\n", B)
print("Element-wise addition:\\n", addition)
print("Element-wise multiplication:\\n", elementwise)
print("Matrix multiplication:\\n", product)
print("Transpose of A:\\n", transpose)
print("Maximum in matrix product:", np.max(product))
print("Minimum in matrix product:", np.min(product))
```

### Q43. Array manipulation challenge with 1–24 reshaped to 4 × 6.

```python
arr = np.arange(1, 25)
matrix = arr.reshape(4, 6)
print("4 × 6 matrix:\\n", matrix)
print("Selected rows (rows 2 and 4, 1-based):\\n", matrix[[1, 3], :])
print("Selected columns (columns 2 and 5, 1-based):\\n", matrix[:, [1, 4]])
print("Transpose (6 × 4):\\n", matrix.T)
print("Row means:", np.mean(matrix, axis=1))
print("Column means:", np.mean(matrix, axis=0))
print("Row sums:", np.sum(matrix, axis=1))
print("Column sums:", np.sum(matrix, axis=0))
print("Unique values:", np.unique(matrix))
```
