## Section C — Hard

### Q26. Create `course_df`; display columns, Student, Student + Score, and add Batch.

```python
course_df = pd.DataFrame({
    'Student': ['Aarav', 'Diya', 'Kabir', 'Meera', 'Ishaan'],
    'Course': ['Python', 'Java', 'SQL', 'Python', 'Java'],
    'Score': [88, 91, 76, 84, 79]
})
print("Columns:", course_df.columns)
print("Student column:\\n", course_df['Student'])
student_score_df = course_df[['Student', 'Score']]
print("Student and Score:\\n", student_score_df)
course_df['Batch'] = '2026'
print("Updated course_df:\\n", course_df)
```

### Q27. Create `travel_df`, add Code, then replace it with a partially indexed Series.

```python
travel_df = pd.DataFrame({
    'Destination': ['Delhi', 'Agra', 'Jaipur', 'Pune', 'Shimla'],
    'Days': [2, 1, 3, 4, 5]
})
travel_df['Code'] = np.arange(5)
print("After np.arange(5):\\n", travel_df)

travel_codes = pd.Series([101, 104], index=[1, 4])
travel_df['Code'] = travel_codes
print("After assigning partially indexed Series:\\n", travel_df)
```

### Q28. Create `school_df`, request missing House, then fill the complete House column.

```python
school_df = pd.DataFrame({
    'Student': ['Aditi', 'Rahul', 'Tanya', 'Arjun'],
    'Class': [9, 10, 9, 10],
    'Section': ['A', 'B', 'C', 'A']
})
school_extended_df = school_df.reindex(columns=['Student', 'Class', 'Section', 'House'])
print("Before assigning House:\\n", school_extended_df)
school_extended_df['House'] = 'Tagore'
print("After assigning House:\\n", school_extended_df)
```

### Q29. Create inventory, add a partially indexed Series column, then delete it.

```python
inventory_df = pd.DataFrame({
    'Item': ['Pen', 'Notebook', 'Bag', 'Bottle'],
    'Quantity': [100, 60, 25, 40],
    'Price': [10, 55, 700, 150]
})
stock_levels = pd.Series(['High', 'Low'], index=[0, 2])
inventory_df['StockLevel'] = stock_levels
print("After adding StockLevel:\\n", inventory_df)
inventory_df = inventory_df.drop(columns=['StockLevel'])
print("After deleting StockLevel:\\n", inventory_df)
```

### Q30. Clipboard table workflow: display, inspect columns, select columns, add and delete a column.

```python
lab_df = pd.read_clipboard()

print(lab_df)

print("Columns:", lab_df.columns)

print(lab_df['Name'])

print(lab_df['Total Marks'])

small_df = lab_df[['Name', 'Total Marks', 'City']]
print(small_df)

lab_df['LabFlag'] = 'Checked'
print(lab_df)

lab_df = lab_df.drop(columns=['LabFlag'])
print(lab_df)
```
